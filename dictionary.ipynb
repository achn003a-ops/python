# dictionary - 

dictionary is a mutable collection of elements that stores the data in key value pairs.
    

```python
student = {
    "name": "Aman Sharma",
    "age": 21,
    "marks": 78,
    "gender": "Male"
}
print(student)
print(type(student))
```

```python
# accessing values of a dictionary

# 1. bracket notation
print(student["name"])

print(student["gender"])

# 2. get()
print(student.get("name"))

print(student.get("marks"))
```

```python
# dictionary can contain any type of data but keys must be immutable and values can be mutable or immutable

student = {
    "roll_no": 1,
    "name": "Akashdeep",
    "marks": [12,13,21,20,19],
    "subjects": {"datascience","python","java","webdev"},
    "address": {
        "zip_code": 112190,
        "city": "Mohali",
        "state": "Punjab"
    },
    "percentage": 56.77,
    "discount": True,
}
print(student)



print(student["address"] ["zip_code"])
print(student.get("address").get("zip_code"))
```

```python
# add a key-value pair to a dictionary

student["email"]="akash77@gmail.com"
print(student)


# adding multiple key-value pairs
student.update({
    "contact": 9888776655,
    "age":21,
})
print(student)
```

```python
# removing elements from a 

# 1. pop() - remove the specified key
student.pop("contact")
print(student)


#2. popitem() - removes the last inserted item
student.popitem()
print(student)


# 3. clear() - empties the dictionary
student.clear()
print(student)
```

```python
# updating  the elements of a dictionary

print(student)

student["percentage"] = 67.77
print(student)
```

```python
# creating empty dictionary 

d = {}
print(d)
print(type(d))
```

```python
# dictionary methods : 

# 1. keys() - return all the keys of a dectionary
print(student.keys())

# 2. values() - returns all the values of a dictionary in list form
print(student.values())

# 3. items() - returns the list of key-value pairs
print(student.items())

# 4. setdefault()
student.setdefault("discount", False)
student.setdefault("contact", 99888776655)
print(student)
```

```python
# traversing on a dictionary

for i in student:
    print(i)
    print(f"{i} : {student.get(i)}")
```

```python
for i in student.values():
    print(i)
```

```python
# dictionary functions

# len()
print(len(student))

# max()
print(max(student))

# min()
print(min(student))

# sorted()
print(sorted(student))
```
