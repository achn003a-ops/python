# list
* A list is an ordered , mutable coollection of elements.
* A list allows to store and manage multiple values inside a single variable


for example, suppose we have to store marks of the five students without a list we will create five variables as :
* student1=85
* student2=92
* student3=76
* student4=88
* student5=95

but with list we can create a single variable and store all these marks inside it

```python
marks = [82, 66, 89, 30, 55, 77]

print(marks)

names = ["Aman", "Harsh", "Priyanka"]
print(names)
```

```python
# list indexing - 

marks = [82, 66, 89, 30, 55, 77]
print(marks[1])

print(marks[-1])


```

```python
# list slicing - 

marks = [82, 66, 89, 30, 55, 77]
print(marks[1:5])
print(marks[:5])
print(marks[2:])
print(marks[::-1])
```

```python
# list mutabilitty - 

marks = [10, 11, 31, 21, 20, 18]

print(marks)

marks[2] = 33
print(marks)
```

```python
# adding new elements to a list - 

nums = [10, 20, 30, 40]

# 1. append() - it is used to add a single element to the end of a list

nums.append(50)
print(nums)

# 2. insert() - it is used to add a element in a list to some specific position

nums.insert(3, 35)
print(nums)

# 3. extend() - can add multiple elements at a time in list

nums.extend([10, 20, 30])
print(nums)
```

```python
# removing elements from a list


nums = [10, 20, 30, 40, 50, 60, 70, 80, 90]

# 1. remove() - removes elements by value
nums.remove(80)
print(nums)

# 2. pop() - removes elements by position
nums.pop(2)
print(nums)

# 3. clear() - empties the list
nums.clear()
print(nums)
```

```python
# list traversal - 

nums = [10, 20, 30, 40, 50, 60, 70, 80, 90]

for i in nums:
    print(i)
```

```python
# list concatination - 

evens = [2, 4, 6, 8]
odds = [1, 3, 5, 7]

print(evens+odds)
```

```python
# other list method / functions :

nums = [10, 20, 30, 40, 50]

# 1. len() - return the length
print(len(nums))

# 2. reverse() - reverse a list
nums.reverse()
print(nums)

# 3. sort() - sorted in ascending
nums.sort()
print(nums)

# to sort in descending : set reverse = True
nums.sort(reverse=True)
print(nums)

# 4. min() - returns minimum value from list
print(min(nums))

# 5. max() - 
print(max(nums))

# 6. index() - returns the index of first occurance of the element 
print(nums.index(10))

# 7. coun() - count the number of occurance of some element in a list
print(nums.count(10))

# 8. sum() -
print(sum(nums))
```

```python
# list cmoprehension - 

l = [10,20,30,40]                         # syntax - new_list = [expression for variable in list]

new_list = [i*i for i in l]
print(new_list)
```

```python
l = [10,20,30,40]

new_list = []
for i in l:
    new_list.append(i*i)

print(new_list)    
```

```python
fruit = ["apple", "mango", "oraneg", "guava", "kiwi", "pineapple"]

fruit_length = []

for i in fruit:
    fruit_length.append(len(i))

print(fruit_length)


# usind list comprehension - 
fruit_len = [len(i) for i in fruit]
print(fruit_len)
```

```python
nums = [10,20,33,42,45,56,76,79,99,70,23]

even_nums = []

for i in nums:
    if i%2==0:
        even_nums.append(i)

print(even_nums)   


# using list comprehension - 
even = [i for i in nums if i%2==0]
print(even)
```
