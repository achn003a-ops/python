# operators

operators are special symbols thaat are used to perform some operations on some values and variables.
The symbol that specifies what operation is going to be applied is called operator.
The values or variables on which the operation to be performed are called operands.


**example**
* a=10
* b=20
* a+b
 *  Here + is the operator, a and b are the operands


#### TYPES OF OPERATORS-


IN python operators are classified in several categories depending on what type of operation do they perform

# 1.Arithematic operators-
The operstors that perform arithematical calculations or mathematical operations are called aritematic operators.
    In pyhton, we have following type of aritematic operators-
* 1. Addition(+)
* 2. Substraction(-)
* 3. Multiplication(*)
* 4. Division(/)
* 5. Floor Division(//)
* 6. Modules(%)
* 7. Exponent(**)

```python
# exapmle arithematic operator

a=100
b=3

#addition
print(a+b)

#substraction
print(a*b)

#multiplication
print(a/b)

#floor division-return only the integer part after dividing two number
print(a//b)

#modules-return remainder that we get after dividing two number
print(a%b)

#exponent
print(a**b)
```

```python
# string with arithematic operators
str1="welcome to"
str2="class"

# addition- string concatination
print(str1+str2)

#multiplication
print(str1*3)
```

```python
# boolean values with arithematic operator - division and modules is not possible
a1=True
a2=False
print(a1+a2)

print(a1-a2)

print(a1*a2)

print(a1**a2)
```

## 2. Comparison operators-
Comparison operator are used to compare two values or varisables.
The result of comparison operators is always either true or false.
We have following comparison operators in python.
 * 1. Greater than(>)
  * 2. Less than(<)
 * 3. Less than equal to(<=)
 * 4. Greater than equal to(>=)
  * 5. Not equal to(!=)
 * 6. equal to(==)

```python
# comparison operator example

a=10
b=100
print(a>b)

print(a<b)

print(a==b)

print(a!=b)

print(a<=b)

print(a>=b)
```

# 3. Assignment Operators
These are used to assign values to variables

1. =
2. additon assignment (+=)
3. Substraction assignment (-=)
4. Multiplication assignment (*=)
5. Division assignment (/=)
6. Floor division assignment (//=)
7. Modules assignment (%=)
8. Exponent Assignment (**=)
   

```python
# example assignment operator

a=10

# addition assignment
#a= a+10
a += 10
print(a)

# substraction assignmenet
a -= 5
print(a)

#multiplicatioin assignment
a *= 6
print(a)

# division assignment

a /=9
print(a)

# floor division assignent
a //= 1
print(a)

#modules assignment
a %= 2
print(a)

#exponent assignment
a **=2
print(a)
```

# 4. Logical operators-
Logical operators are used to combine multiple conditions written using comparison operators to return or produce a single output.
We have three logical operators in python:
1. AND
2. OR
3. NOT

```python
#example logical operator

a=10
b=100

#AND
print(a>100 and a>0)

print(a<100 and a>0)
print(a<100 and a>0 and a>b)

#OR
print(a<100 or a>0 or a>b)

#NOT
print(not a>b)
```

# 5. Membership operators
These are used to check whether something belongs to a sequence or not.
There sre two membership operators:
1. in
2. not in

```python
# example membership operator

msg = "welcome to python"
print('hi' in msg)
print('w' in msg)
print('W' in msg)

# not in
print('W' not in msg)
print('python' not in msg)
```

# 6. Identity Operators
Identity operators are used to check whether two objects refers to the same memory
1. is

```python
# example identity operator

a=10
b=10
print(a is b)

str1='hello'
str2='hello'
print(str1 is str2)
print(str1 is not str2)
```
