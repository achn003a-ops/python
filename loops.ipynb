# for loop - 
for loopis used to iterate(visit every element and perform actions) over sequence(collection of elements)/iterables

 It works in string, sets, tuples

```python
# in - it is a membership operator
# in this char is an variable


msg = "hello"

for char in msg:
    print(char + "00")
```

```python
# find the sum of the even numbers lying between 1 and 20 and store in the variable named even_sum
# and the sum of all odd numbers in variable names odd_sum
# do both of the in a single loop

even_sum = 0
odd_sum = 0

for i in range(1, 21):
    if i%2 == 0:
        even_sum += i
    else:
        odd_sum += i
print("sum of even number: ", even_sum)
print("sum of odd number: ",odd_sum)
   
    
```

```python
# ask the user to input price of 5 different items
#  ignore all the negative values provided
#  find sum and average of all the positive ones


total = 0
count = 0
for i in range(1, 6):
    price = int(input("enter a item: "))
    clear_output()
    if price < 0:
        continue
    print(price) 
    total += price
    count += 1
    avg = total/count
    
print("total price: ", total)
print("average of price: ",avg)
```
