# typecasting - 
The process of converting the datatypes of the varibles from one to another is called typecasting.
In python, typecasting is of two types -
1. implicit typecasting -
The typecasting done by the pythons interpreter on its own without any intervention of the programmar is called implicit typecasting

2. Explicit typecasting -
This type of typecasting is done by the programmar using some in-built functions to manually convert the datatypes of the variables.

```python
# implicit  - 

res = True
num = 10
add = res + num
print(type(add))
print(add)
```

```python
v1 = 10.90
v2 = 10

add = v1+v2
print(add)
print(type(add))
```

```python
# 1. int(var) - that converts the data types of any variable to int

float_val = 10.98
str_val = '19'
bool_val = True
print("before conversion")
print(type(float_val))
print(type(str_val))
print(type(bool_val))

float_val = int(float_val)
str_val = int(str_val)
bool_val = int(bool_val)

print("-----------\nAfter conversion")
print(type(float_val))
print(type(str_val))
print(type(bool_val))

print(float_val + str_val + bool_val)
```

```python
# 2. str()

float_val = 10.98
int_val = 19
bool_val = True
print("before conversion")
print(type(float_val))
print(type(int_val))
print(type(bool_val))

float_val = str(float_val)
int_val = str(int_val)
bool_val = str(bool_val)

print("-----------\nAfter conversion")
print(type(float_val))
print(type(int_val))
print(type(bool_val))

```

```python
# 3. float -

int_val = 10
str_val = '19'
bool_val = True
print("before conversion")
print(type(int_val))
print(type(str_val))
print(type(bool_val))

int_val = float(int_val)
str_val = float(str_val)
bool_val = float(bool_val)

print("-----------\nAfter conversion")
print(type(int_val))
print(type(str_val))
print(type(bool_val))

```

```python
# 4. bool() - any non-zero integer, either positive or negative, when converted to bool, always return truw. only zero is converted to false

num = 10
num = bool(num)
print(num)
print(type(num))


num2 = 0.0
num2 = bool(num2)
print(num2)
print(type(num))
# only 0.0 is converted to false, every other float value gives true

string = ''
string = bool(string)
print(string)
print(type(string))
# empty string '' returns false when converted to boolean and every non-empty string is converted to true
```
