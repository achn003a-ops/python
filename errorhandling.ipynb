# Error - 
An error is an problem that prevents the program to execute normally.

## Types of error in Python - 
1. Syntax Error - An error that occurs when we wrtie something that doesn't follow the basiic python syntax.
2. Logical Error - when a program executes normally, but the output is not as expected due to the incorrect logic written in the code.
3. Runtime Error / Exceptions - The error that occurs during the execution of the program.

# Exception Handling in Python - 
Exception handling is a mechanism used to detect and handle runtime exceptions so that program doesn't terminate unexpectedly.

There are three main things used to handle these exceptions in python 
1. try
2. except
3. else
4. finally

```python
# some of the built-in python exceptions examples that can be handled easily with exception handling :

num1 = 100
num2 = 0

print("Dividing........")
print(num1/num2)
```

```python
num1 = 100
num2 =  0

try:
    res = num1/num2

except:
    print("Cant't divide a number by zero")

else:
    print("Division of num1 and num2 is", res)


print("hello")
```

```python
num1 = input("Enter number 1")
num2 = input("Enter number 2")

try: 
    num1 = int(num1)
    num2 = int(num2)
    res = num1/num2
except ValueError:
    print("Invalid data to be converted to integer")
except ZeroDivisionError:
    print("Number cannot be divided by 0")
else:
    print("Division of two numbers is :", res)
    
    
```

```python
num1 = input("Enter number 1")
num2 = input("Enter number 2")

try: 
    num1 = int(num1)
    num2 = int(num2)
    res = num1/num2
except ValueError as e:
    print(e)
except ZeroDivisionError as e:
    print(e)
else:
    print("Division of two numbers is :", res)
finally:
    print("Code execution is completed")
```

```python
# write a program that asks user for a list index and divides 100 by selected elements. Handle all possible exception.

listt = ["hello", 21, 13, 0, "adwd"]

user = int(input("Select the list index"))

try:
    index = 100/listt[user]
    
except ValueError as e:
    print(e)

except ZeroDivisionError as e:
    print(e)

except IndexError as e:
    print(e)

except TypeError as e:
    print(e)


else:
    print("Division of two numbers is :", index)

finally:
    print("Execution is completed")
```

```python
acc_bal = 89700
atm_pin = 9090

user_atm_pin = int(input("Enter your atm pin"))
if user_atm_pin == atm_pin:
    withdraw_amt = int(input("Amount to withdraw :"))
    try:
        if withdraw_amt > acc_bal:
            raise ValueError("Insufficient Amount Balance")
    except ValueError as e:
        print(e)
    else:
        print("Transaction Successful. Available Balance :", acc_bal-withdraw_amt)

else:
    print("Incorrect pin provided")
```

```python
class UserNameError(Exception):
    pass

class PasswordError(Exception):
    pass


try:

    username = input("Enter Username :")
    password = input("Enter password :")

    
    if len(username) < 4: 
        raise UserNameError("Username must contain atleast 4 characters")

    if username.isspace():
        raise UserNameError("Username cannot contain only spaces")

    if username.isdigit():
        raise UserNameError("Username cannot cantain only digits")
        

    if len(password) <8 :
        raise PasswordError("Password must contain atleast 8 characters")

    if password.isalnum():
        raise PasswordError("Password cannot contain only digits")




except UserNameError as e:
    print(e)
    
except PasswordError as e:
    print(e)

else:
    print("Usernaem and Password are correct")
    
```
