```python
class A:
    def __init__(self, x, y):
        
        self.x = x
        self.y = y

    def display(self):
        print("x: ", self.x)
        print("y: ", self.y)

class B(A):
    def __init__(self, x, y, z):
        super().__init__(x,y)
        self.z =z

    def display(self):
        super().display()
        print("z: ", self.z)

b1 = B(10, 20, 30)
b1.display()

    
    
```

```python
class Person:
    def __init__(self, name, age, gender):
        self.name = name
        self.age = age
        self.gender = gender

    def display(self):
        print("Name :", self.name)
        print("Age :",self.age)
        print("Gender :", self.gender)

class Student(Person):
    def __init__(self, name, age, gender, college, course, fees):
        super().__init__(name, age, gender)
        self.college = college
        self.course = course
        self.fees = fees

    def display(self):
            super().display()
            print("College Name :", self.college)
            print("Course Name :", self.course)
            print("Course Fees :", self.fees)



student1 = Student("jon", 21, "male", "SD college", "BCA", 50000)
student2 = Student("Alice", 20, "female", "SD College", "Btec", 60000)



student1.display()
student2.display()

        
```

```python
# Types of inheritence - 
# 1. Single Inheritence - A single class inherits from single parent class.
# eg. A ---> B

# 2. Multiple Inheritence - A child inherits from multiple parent class.
# eg. Class A & class B ----> Class C
        

# 3. Multilevel inheritence -

# eg. A --> B --> c

# 4. Hierarichal Inheritence - Multiple child class are created from a single class.
# eg. A --> B and C
```

```python
# Multiple Inheritence example - 

class Academic:
    def __init__(self, marks):
        self.marks = marks

    def display(self):
        print("Marks : ", self.marks)

class Sports:
    def __init__(self, report):
        self.report = report
    def display(self):
        print("Report : ", self.report)

class Student(Academic, Sports):
    def __init__(self, marks, report, name):
        Academic.__init__(self, marks)
        Sports.__init__(self, report)
        self.name = name

    def display(self):
        Academic.display(self)
        Sports.display(self)
        print("Name :", self.name)

s = Student(50, "A", "jon")
s.display()
```

```python
# Multilevel Inheritence example - 

class Academic:
    def __init__(self, marks):
        self.marks = marks

    def display(self):
        print("Marks : ", self.marks)

class Sports(Academic):
    def __init__(self, marks, report):
        super().__init__(marks)
        self.report = report
        
    def display(self):
        super().display()
        print("Report : ", self.report)

class Student(Sports):
    def __init__(self, marks, report, name):
        super().__init__(marks, report)
        self.name = name

    def display(self):
        super().display()
        print("Name :", self.name)

s = Student(50, "A", "jon")
s.display()
```

```python
# Hierarichal Inheritence example - 

class Student:
    def __init__(self, name , course):
        self.name = name
        self.course = course

    def display(self):
        print("Name :", self.name)
        print("Course :", self.course)

class Academic(Student):
    def __init__(self, name, course, marks):
        super().__init__(name, course)
        self.marks = marks

    def display(self):
        super().display()
        print("Marks :", self.marks)

class Sports(Student):
    def __init__(self, name, course, report):
        super().__init__(name, course)
        self.report = report

    def display(self):
        super().display()
        print("Report :", self.report)

s = Student("jon", "Btech", 50, "A")
s.display()
```
