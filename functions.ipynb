```python
# ask user to input two numbers and create 7 different functions and print 7 arithematic operations on those two numbers

num_1=int(input("enter a number : "))
num_2=int(input("enter a number : "))

def arithematic():
    add=num_1+num_2
    print(add)

def subtraction():   
    sub=num_1-num_2
    print(sub)

def  multiplication():     
    mul=num_1*num_2
    print(mul)

def division():    
    div=num_1/num_2
    print(div)

def floordivision():    
    floor=num_1//num_2
    print(floor)

def modulus():    
    modulus=num_1%num_2
    print(modulus)

def exponent():    
    exponent=num_1**num_2
    print(exponent)
    

arithematic()   
subtraction()
multiplication()
division()
floordivision()
modulus()
exponent()
    
```

```python
def add():
    
    num1 = int(input("enter number 1 "))
    num2 = int(input("enter number 2"))
    res = num1 + num2
    print("Result", res)
    return num1, num2, res


# whenever we try to return multiple values from a function, a tuple of those values is created, 
# that means the statement we can return on;y a single object from a function remain valid

result=add()
num1, num2, res = result

print(num1, num2, res)


# if we want to access values of a tuple returned by the function, we can do tuple unpacking
```

```python
# * - args - arguements

def studentInfo(name,age, *marks):                 
    print("Student Name: ", name)
    print("Student Age: ", age)
    marks=list(marks)
    print("Student Marks: " , marks)

studentInfo("john", 22, 25, 24, 21)    
```

```python
# **kwargs - expecting a dictionary

def studentInfo(**info):
    print(info)

studentInfo(name="john", age=22, marks=[10,11,12])    
```

```python
# keyword arguements, positional arguement

def student(name, age, contact, course):
    print(name, age, contact, course)

student("john", 21, 9988776655, "Btech")    

student(age=21, contact=7766554433, name="john", course="bca")
```
