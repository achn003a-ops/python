```python
# constructor - 
# 1. Constructor is a function that is automatically invoked at the time of object creation.
# 2. Everytime an object is created from a class, the constructor function gets automatically called.
# 3. The name of constructor function is fixed : init
# 4. The constructor function is used to assign the values to the attributes for each object.
```

```python
# Example - 

class student:
    def display(self, name, age):
        print(name, age)

s1 = student()
s1.name = "Rahul"
s1.age = 21
s1.display(s1.name, s1.age)

s2 = student()
s2.name = "Priya"
s2.age = 22
s2.display(s2.name, s2.age)

# in these examples assigning the values to the objects required repetition and we need to assign the values one by one
# to avoid this hustle, constructor option is the bes
```

```python
# example using constructor : 

class student:
    def __init__(self):
        print("constructor function called")
    def display(self):
        print("Hello")

s1 = student()
s2 = student()

# eventhrough we are not calling the __init__function, still it is beign called, this is speciallity of constructor function.
    
```

# Class Variables - 
class variables and instance variable

# 1. Instance Variable - 
It belongs to a particular object. Every object can have differnt values for instance variable. They are usually created an assigned values using self keywrd.
# 2. Class Variable - 
It belongs to the class itself and is shared by all the objects. The class variable are defined directly inside a class outside any methods. Same value is available to all the objects. CLass variables are useful for common information.

```python
# class variables - 

class student:
    college_name = "ABC College"
    def __init__(self, name, age, marks):
        self.name = name
        self.age = age
        self.marks = marks
    def display(self):
        print(self.name, self.age, self.marks,self.college_name, sep="\n")

s1= student("Rahul", 21, 97)
s2 = student("Priya", 22, 99)

s1.display()
s2.display()
```

# Instance Method - 
It works with the data of paritcular object. It takes self as first parameter and instance method can access and modify instance variable. It is always called using an object.

# Class Method -
It works with data that belongs to the class rather than a particular object. It is created using (@classmethod). It takes class as the first paramter. A class method can access and modify class variables. It can be called using class or an object.

```python
class company:
    company_name="Netmax"
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
    def display(self):
        print("Employee Name:", self.name)
        print("Employee Salary:", self.salary)
        print(self)
    @classmethod
    def display_comp(self):
        print("Company Name:", self.company_name)
        print(self)

e1 = company("Practiksha", 42000)
e2 = company("Divyansh", 43000)

e1.display()
e2.display()

e1.display_comp()
```
