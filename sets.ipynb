# sets - 
Sets are unordered collection of unique elements.

# Charactersitics of set
* Unordered
* does not contain duplicates
* mutabele
* con contain different data types
* elements must be immutable

```python
# creating a set - {} are used to create a set

s = {1,2,3,4}

# can contain different data types
s2 = {1, "Rahul", 24, True, 67.88}
print(s2)
```

```python
# set do not contain duplicate values

s = {10,10,20,30,40,50,10,10}
print(s)
```

```python
# set itself is mutable, but the elements of set is immutable

s= {"Rahul", [10,20,30,40,]}   // immutbale element dont work

s= {"Rahul", (10,20,30,40,)}
print(s)
```

```python
# creating empty set

# r = {}
# print(r)

# print(type(r))


# correct method of creating empty set

s = set()
print(s)

print(type(s))
```

```python
# adding elements to a set

s = {10,20,30,}

# 1. add() - add single element to a set
s.add(40)
print(s)

# try to add multiple element in set using add will raise an error
#s.add(10,20)
#print(s)


# 2. update() - can insert multiple elements in a set at a time 
s.update((100,200,300))
print(s)
```

```python
# removing elements from a set 

s = {10,20,30,40,50,60}
# 1. remove() - remove the specified element from a set

s.remove(10)
print(s)

# 2. pop() - removes a random element
s.pop()
print(s)

# 3. clear() - empties a set
s.clear()
print(s)

# s.remoce(100)
#print(s)

# 4. discard()
s.discard(100)
print(s)
```

```python
# set operation - 

s1 = {10,20,30,40,50,60,70}
s2 = {50,60,70,80,90,100}

# 1. union() - combines two sets
print(s1.union(s2))
print(s1 | s2)           # (| - or operator)

# 2. intersection() - returns common elements from two sets
print(s1.intersection(s2))
print(s1 & s2)         # (& - AND operator)

# 3. difference () - return the values from set a that are not present in set b
print(s1.difference(s2))
print(s2.difference(s1))

# with - operator
print(s1-s2)
print(s2-s1)


# 4. symmetric_difference()
print(s1.symmetric_difference(s2))
print(s1 ^ s2)

# 5. issubset
print(s1.issubset(s2))

# 6. superset
print(s1.issuperset(s2))

```

```python
s1 = {10,20,30,40,50}
s2 = {10,20}

print(s1.issubset(s2))
print(s2.issubset(s1))

print(s1.issuperset(s2))
```

```python
# 7. isdisjoint
a= {1,2,3,4}
b= {'a','b','c','d'}

print(a.isdisjoint(b))
```

```python
# converting other data types to sets

l = [10,20,30,40,50,60,30,20]

newlist = set(l)
print(newlist)
```

```python
# converting tuple to a set
t = (10,20,30,40,50,10,20,30)
newt = set(t)
print(newt)
```

```python
# traversing over a set

s = {10,20,30,40,50}

for i in s:
    print(i)


print(len(s))    
print(max(s))
print(min(s))
print(sum(s))
```

```python
# set comprehension - 

s1 = {1,2,3,4,5,6}
s2 = {i*i for i in s1}
print(s2)
```

```python
s3 = {11,12,13,14,15,16,17,18}

s4 = {i for i in s3 if i%2==0}
print(s4)
```
