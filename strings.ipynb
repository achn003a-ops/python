```python
# string - 

msg = "python"
print(msg)
print(msg[2])
```

```python
# indexing - Each string has a position called index
# python uses 0 based indexing that means the index start from 0 not from 1
# string - P Y T H O N
#index -  0 1 2 3 4 5 

msg ="python"
print (msg[-1])
```

```python
# string slicing - It is used to extract a part of string

# syntax-
# str_name[start:stop]

# example

msg = "how are you"

print(msg[1:6])

# start index is inclusive but stop index is excluded

# if we want to slice a string from starting to some specific position
print(msg[0:7])
print(msg[:7])
print(msg[3:])

# slicing with step value
print(msg[0:7:2])

# slicing in reverse order
print(msg[::-1])
```

```python
# traversing a string
for ch in msg:
    print(ch)
```

```python
# case-conversion methods - 

msg = "Python is easy"

# 1. upper()
print(msg.upper())

# 2. lower()
print(msg.lower())

#3. Capitalize
print(msg.capitalize())

# 4. title()
print(msg.title())

# 5. swapcase()
print(msg.swapcase())
```

```python
# string searching methods :

# 1. find()
msg = "Python is easy"
print(msg.find("a"))

print(msg.find("u"))

# 2. search()
print(msg.index("is"))

# print(msg.index("are"))


# When we try to look for a value that is not present in string, find returns -1 but index raises an error

# 3. count()
print(msg.count("y"))


# 4. startswith()
print(msg.startswith("P"))

# 5. endswith()
print(msg.endswith("n"))
```

```python
# removing spaces from a string

# strip()
string = "                   Hello         "
print(string)

print(string.strip())


# lstrip()
print(string.lstrip())


# rstrip()
print(string.rstrip())
```

```python
# replacing in string :

msg = "python is easyyyy"

print(msg.replace("y","iii"))

print(msg.replace("y","iii",3))
```

```python
# splitting a string :

msg = "python"
print(msg.split())
```

```python
# string validation methhod :

msg = "Python"

# 1. isupper()
print(msg.isupper())
print("PYTHON".isupper())

# 2. islower()
print("python".islower())

# 3. isalpha()
print("python".isalpha())
print("How are you".isalpha())

# 4. isdigit()
print("123a".isdigit())

# 5. isalnum()
print("123a".isalnum())

# 6. isspace()
print("  ".isspace())
```

```python
# Ask the user to enter a string and count:
#     - Total characters
#     - Uppercase characters
#     - Lowercase characters
#     - Digits
#     - Spaces

string = input("enter a number :")

total = len(string)
uppercase = 0
lowercase = 0
digit = 0
space = 0



for ch in string:
    
    if ch.isupper():
        uppercase += 1
    elif ch.islower():
        lowercase += 1
    elif ch.isdigit():
        digits += 1
    elif ch == " ":
        spaces += 1

print("Total:", total)
print("Uppercase:", uppercase)
print("Lowercase:", lowercase)
print("Digits:", digits)
print("Spaces:", spaces)





```

```python
# 3. Ask the user for a username.
#     The username is valid only if:
#      - It contains only alphabets and numbers.
#      - It contains at least 5 characters.

user = input("enter a username :")
if user.isalnum() and len(user) >= 5:
    print("Valid Username")
else:
    print("Invalid Username")
```

```python
# Ask the user for an email address.
#     Check whether:
#     - It ends with "@gmail.com".
#     - It contains "@".


email = input("enter an email : ")
if email.endswith("@gmail.com") and "@" in email:
    
    print("valid email")
else:
    
    print("invalid email")
    
```
