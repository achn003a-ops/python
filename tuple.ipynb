# tuple - 
Tuple is an ordered collection of immutable elements in python.


```python
# creating a tuple

t = 12,13,14,15
print(t)
print(type(t))
```

```python
# indexing in tuple - 

# positve indexing
print(t[0])

# negative indexing
print(t[-1])
```

```python
# slicing
print(t[0:4])
print(t[:4])
print(t[2:])
print(t[::-1])
```

```python
# creating empty tuple - 

t =()
print(t)
print(type(t))

t = tuple()
print(t)
print(type(t))
```

```python
# creating a single item 

t= (12,)
print(t)
print(type(t))
```

```python
# tuple methods : 

t=(10,20,30,40,50,60)

print(t.index(10))

print(t.count(20))
```

```python
# tupel functions - 

t=(10,20,30,40,50,60)

# 1. len()
print(len(t))

# 2. max
print(max(t))

# 3. min
print(min(t))

# 4. sorted()
print(sorted(t))

# 5. sum()
print(sum(t))
```

```python
# tuple concatenation - 

t1 =(10,20,30)
t2 =(40,50,60)

print(t1+t2)
```

```python
# tuple packing - 

t = 78,89,90

a ="mango","apple","mango"
print(t,a)
print(type(t), type(a))
```

```python
# tuple unpacking - 

fruit,color,vegetable,name=("mango","red","potato","anjali")
print(fruit)
print(color)
print(vegetable)
print(name)
```

```python
t = (10,20,30,40,50)

m1, m2, m3, m4, m5 = t
print(m1,m2,m3,m4,m5, sep ="\n")
```

```python
t = ("mango", "red", "orange", "pink", "yellow")

fruit, *color = t
print(fruit)
print(color)
```

```python
t = (10,20,30, (100,200,300))
print(t)

print(t[3][1])
```
