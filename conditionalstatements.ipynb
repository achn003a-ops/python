```python
# write a program that accepts users age and check whether they belong to one of the following - 

# a. child(0-12)
# b. teenager(13-19)
# c. adult(20-59)
# d. senior citizen( 60 and above)

age = int(input("enter your age: "))

if age <= 12:
    print("you are a child")
elif age >=13 and age <= 19:
    print("you are a teenager")
elif age >=20 and age <= 59:
    print("you are an adult")
elif age >= 60:
    print("you are a senior citizen")
```
