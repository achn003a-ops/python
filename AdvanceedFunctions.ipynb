```python
# What is MRO?
# MRO is the order in which Python searches for a method or attribute when working with inheritance.
# It becomes especially important with multiple inheritance.
```

```python
# global
# global is used inside a function when you want to modify a variable that belongs to the global scope.

# nonlocal
# nonlocal is used inside a nested function when you want to modify a variable belonging to the outer

# global → outside all functions
# nonlocal → outside the current nested function
```

```python
# why are functions called first class object in python

# In Python, functions are called first-class objects because you can treat a function like any other value/object.
# That means you can store a function in a variable, pass it as an argument, return it from another function, and store it in data structures.

# Why is this useful?
# This feature allows Python to support things like:
#Callbacks
# Decorators
# Higher-order functions
# Functional programming
# Passing behavior dynamically
```

```python
# Higher order function - A function is called a higher order function if it satisfies any of these two conditions :-
# 1. It is returning a function
# 2. It is accepting another function as an arguement
```
